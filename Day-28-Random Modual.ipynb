```python
import numpy as np
```

```python
# Generate Random Values From 0 to 1 in Float

print('1 Random Value              :', np.random.rand())          # 1 Random Value
print('\n')

print('5 Random Value              :', np.random.rand(5))         # 5 Random Value
print('\n')

print('Random Values in 3*3 Matrix :\n', np.random.rand(3, 3))     # Random Values in 3*3 Matrix 

print('\nRandom Values in 2*3*4 Matrix :\n', np.random.rand(2, 3, 4))  # Random Values in 2*3*4 Matrix 

# 2 (The First Dimension): There are 2 main blocks (often thought of as "matrices" or "pages").
# 3 (The Second Dimension): Each of those 2 blocks contains 3 rows.
# 4 (The Third Dimension): Each row contains 4 columns.
```

```python
# Generate Random Values From 0 to 1 in Integer

print('Random Values From One to 10            :', np.random.randint(1, 10))
print('\n')

print('5 Random Values From One to 10           :', np.random.randint(1, 10, 5))      
print('\n')

print('3*3 Matrics Random Values From One to 10: \n', np.random.randint(1, 10, (3, 3))) 

```

```python
# Get Random Choice From Dataset
data = np.array([1, 2, 3, 4, 5, 6])

print('Single Random Value Select From Provided Dataset :', np.random.choice(data))
print('Double Random Value Select From Provided Dataset :', np.random.choice(data, 2))
```

```python
# Shuffle Data Randomly
data = np.array([10, 20, 30, 40, 50, 60])

np.random.shuffle(data) 

print(data)
```

```python
# Shuffle Data Randomly and Make New Copy in System
data = np.array([10, 20, 30, 40, 50, 60])

shuffle_data = np.random.permutation(data) 

print(shuffle_data)
```

```python
# Fixed Values Every Time Run 
np.random.seed(42)
print(np.random.randint(1, 10, 5))
```
