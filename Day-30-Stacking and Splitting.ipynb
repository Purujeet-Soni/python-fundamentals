```python
import numpy as np
```

## Stacking

```python
# Array
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
```

### Horizontal Stacking

```python
print('Horizontal Stacking :\n', np.hstack((arr1, arr2))) # Same No of Rows and Dimension
```

### Vertical Stacking

```python
print('Vertical Stacking :\n', np.vstack((arr1, arr2)))
```

### Columns Stacking

```python
print('Column Stacking :\n', np.column_stack((arr1, arr2)))
```

### Rows Stacking

```python
print('Row Stacking :\n', np.row_stack((arr1, arr2)))
```

## Split

```python
# Data
Arr_1 = np.array([10, 20, 30, 40, 50, 60])
Arr_2 = np.array([[10, 20, 30, 100], [40, 50, 60, 200]])
```

#### Split Array In 3 Equal Parts

```python
print('Array Divide In 3 Equal Parts :\n', np.split(Arr_1, 3))      # np.split(array, no_in_Split)
```

#### Horizontally Split Array In 3 Equal Parts

```python
print('Array Horizontally Divide In 4 Equal Parts :\n', np.hsplit(Arr_2, 4))      # np.hsplit(array, no_in_Split)
```

#### Vertically Split Array In 3 Equal Parts

```python
print('Array Vartically Divide In 4 Equal Parts :\n', np.vsplit(Arr_2, 2))      # np.vsplit(array, no_in_Split)
```
