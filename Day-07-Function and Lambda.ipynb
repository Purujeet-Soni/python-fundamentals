```python
# Greet Function
def greet():
    print("Hello")

greet()
```

```python
# Square Function
def square(x):
    print(x*x)

square(12)
```

```python
# Addition Function
def Addition(a, b):
    print(a + b)

Addition(23, 20)
```

```python
# Addition Function for 3
def Addition(a, b, c):
    print(a + b + c)

Addition(23, 20, 23)
```

```python
# Arguments Addtion Function Multilple Numbers
def Addition(*a):  # Used for Multiple Input Parameter / Unknown No of Parameter
    print(sum(a))
    print(a[1])    # It Store in List so we can Print By Indexing Them

Addition(2, 23, 300000)
```

```python
# Keyword Arguments Multiple Key : Value Pair
def show(**data):
    print(data)  # Store in Dictionary Formate

show(name = "Purujeet Soni", level = 69)
```

```python
# Return
def Addition(a, b):
    return a + b

c = Addition(23, 7)
d = Addition(20, 5)
print(c)
print(d)
```

```python
# Lambda Function

# Single Variable Lambda Function

cube = lambda x : x**3                   # variable = lambda Arguments : Expresions
print(cube(5))                          # print(variable(values))



# Multi Variable Lambda Function

add = lambda x, y : x + y
print(add(2, 3))
```

```python
# Arguments in Lambda
add_all = lambda *x: sum(x)
print(add_all(1, 2, 3, 4, 5))   # 15
```
