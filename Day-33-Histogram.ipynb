### Library

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

#### Data

```python
data = [10, 11, 20, 25, 18, 11, 20, 25, 10]
```

### By plt

#### Simple

```python
plt.figure(figsize=(12, 5))
plt.hist(data)
plt.xticks(range(10, 28))
plt.show()
```

#### By Specific Bins 

```python
plt.hist(data, bins = 7)
plt.show()
```

#### By Specific Releative Width of Bar  

```python
plt.hist(data, bins = 17, rwidth=0.5)
plt.show()
```

#### By Specific Color

```python
plt.hist(data, color = 'gold', edgecolor='orange')
plt.show()
```

### By Seaborn

#### Simple

```python
plt.figure(figsize=(6, 3))
sns.histplot(data)
plt.show()
```

####  By Specific Bins

```python
sns.histplot(data, bins=4)
plt.show()
```

```python
# Plot bins that Out of Data Range
sns.histplot(data, bins=40)
plt.show()
```

#### Plot KDE Line With Histogram

```python
sns.histplot(data, kde=True)
plt.show()
```
