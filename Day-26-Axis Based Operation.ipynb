```python
import numpy as np
```

# 1. Accross the Columns And Rows (Sum)

```python
arr = np.array([range(1, 3), range(3, 5)])
arr
```

```python
# Sum All Elements
all_sum = arr.sum()
print(all_sum)
```

```python
# Sum All Rows Elements
row_sum = arr.sum(axis = 1)
print(row_sum)
```

```python
# Sum All Column Elements
row_sum = arr.sum(axis = 0)
print(row_sum)
```

# 2. Accross the Columns And Rows (Min, Max)

```python
# Find Maximum Accross the Rows And Columns
max_accross_row = arr.max(axis=1)
print('Max Accross Row :\t', max_accross_row)

max_accross_Columns = arr.max(axis=0)
print('Max Accross Columns :\t', max_accross_Columns)
```

```python
# Find Minimum Accross the Rows And Columns
min_accross_row = arr.min(axis=1)
print('Max Accross Row :\t', min_accross_row)

min_accross_Columns = arr.min(axis=0)
print('Max Accross Columns :\t', min_accross_Columns)
```

# 3. Return Index Number

```python
# This Return Index of Minimum Array Element Row Wise
print(arr.argmin(axis=1))
```

```python
# This Returaxn Index of Maximum Array Element Column Wise
print(arr.argmax(axis=0))
```

```python
# This Returaxn Index of Array
arr = np.array([10, 20, 30, 40, 50])

index = np.where(arr == 30)
print(index)
```

# 4. 3D Array Matric Operations Sum by Depth, Row, Columns

```python
arr = np.array([range(1, 13)])
arr = arr.reshape(2, 2, 3)
arr
```

```python
# 3D Array Depth Wise Sum (Over Lap At Each Other and Sum)


print(arr.sum(0))

# Same
# print(np.sum(arr, axis = 0))
```

```python
# 3D Array Row Wise Sum (Stack both cut and Sum)


print(arr.sum(1))

# Same
# print(np.sum(arr, axis = 1))
```

```python
# 3D Array Columns Wise Sum (Vertically Fold and Sum)


print(arr.sum(2))

# Same
# print(np.sum(arr, axis = 2))
```

# 5. Keep Dimension as it is Not Change

```python
row_sum = np.sum(arr, axis=1, keepdims=True)

print(row_sum)
print()

col_sum = np.sum(arr, axis=2, keepdims=True)
col_sum
```
