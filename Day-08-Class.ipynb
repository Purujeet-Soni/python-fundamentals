```python
# How to access data using objects
# Think of a class as a "design" and objects as "real things" made from it.

class Students:
    # They belong to the class and are shared by all objects created from this class.

    name = "Purujeet Soni"   # Student name
    Age = 23                # Student age


# Now we create objects of the Students class.
# An object is a real instance created from the class blueprint.

d1 = Students()   # d1 is an object of Students class
d2 = Students()   # d2 is another object of Students class


# Using the object, we can access class variables using dot (.) operator.
# object_name.variable_name

print(d1.name)   # Accessing 'name' using object d1
print(d2.Age)    # Accessing 'Age' using object d2


# OUTPUT:
# Purujeet Soni
# 23
```

```python
# How changing values using an object works
class Students:
    name = "Purujeet Soni"   # Default name for every student
    Age = 23                # Default age for every student


# each object can store its own data.
d1 = Students()   # First object
d2 = Students()   # Second object


# Changing values using objects
# When we assign a value using an object,
# Python creates an INSTANCE VARIABLE for that object only.
# This change does NOT affect the class or other objects.

d1.name = "Ravan"   # d1 now has its own 'name'
d2.Age = 486        # d2 now has its own 'Age'


# Printing values
# --------------------------------
# d1.name → Python finds 'name' in d1 first
# d2.Age → Python finds 'Age' in d2 first

print(d1.name)   # Output: Ravan
print(d2.Age)    # Output: 486


# IMPORTANT NOTE:
# d2.name is still "Purujeet Soni"
# d1.Age is still 23
# because we did NOT change those values for them.
```

```python
class Students:
    name = "Purujeet Soni"
    Age = 23

# Two objects created from the same class
s1 = Students()
s2 = Students()

# Instance-level assignment
# This creates instance variables ONLY inside s1
# and overrides the class variables for s1

s1.name = "Ravan"
s1.Age = 486


# Attribute lookup rule:
# Python checks in this order:
# 1. Instance (object) dictionary
# 2. Class dictionary
# 3. Parent classes (if any)

print(s1.name, s1.Age)
# Output comes from s1's instance variables

print(s2.name, s2.Age)
# No instance variables in s2
# Values are read directly from the class variables
```

```python
# Focus:
# - How instance variables override class variables
# - What 'self' represents inside a class method

class Student:
    name = "Raj"      # Class variable (default value)

    def show(self):
        # 'self' refers to the object that calls this method
        # Here, self.name means:
        # 1. Look for 'name' inside the calling object
        # 2. If not found, look inside the class
        print(self.name)


# Creating an object of Student
s = Student()


# Creating an instance variable 'name' for object s
# This overrides the class variable ONLY for this object
s.name = "Purujeet Soni"


# Method call using the object
# Python internally converts this to:
# Student.show(s)

s.show()   # Output: Purujeet Soni
```

```python
# PROGRAM: Using constructor (__init__) to initialize objects
# ----------------------------------------------------
# Focus:
# - What __init__ is
# - How object-specific data is created
# - How values are stored separately for each object
# ----------------------------------------------------


class Students:
    def __init__(self, name, marks):
        # __init__ is a special method called automatically
        # when a new object of the class is created

        # 'self' refers to the current object being created

        # These are INSTANCE VARIABLES
        # Each object gets its own copy of these variables

        self.name = name     # Assign constructor parameter 'name' to object variable
        self.marks = marks   # Assign constructor parameter 'marks' to object variable


# Creating first object
# Values "Ram" and 19 are passed to __init__
# Python internally does: Students.__init__(s1, "Ram", 19)

s1 = Students("Ram", 19)


# Creating second object
# Values "Shayam" and 20 are passed to __init__
# A separate memory space is created for s2

s2 = Students("Shayam", 20)


# Accessing instance variables using objects
# Object.variable_name

print(s1.name, s1.marks)   # Output: Ram 19
print(s2.name, s2.marks)   # Output: Shayam 20
```

# Encapsulation
Wrapping data and methods together and protecting data from direct misuse.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # private variable define by '__' double slash

    def withdraw(self, amount):
        self.__balance -= amount

    def  get_balance(self):
        return self.__balance


t = BankAccount(3000)
t.withdraw(1000)
print(t.get_balance())
```

# Inheritance (Reusing Code)
One class uses properties of another class,  
e.g --> Child inherits surname, house, habits

```python
class Animal:
    def eat(self):
        print("Eating")

    def see(self):
        print("See Color")

class Dog(Animal):
    def bark(self):
        print("Barking")


d = Dog()
d.eat() # Now Dog Can Inheret Properties of Animal of Eating
d.see() # Now Dog Can Inheret Properties of Animal of See
```

# Polymorphism (Many Forms)    
One function name, different behavior,   
eg -->  Remote button Same button Different action for TV, AC, Music system

```python
class Dog:
    def sound(self):
        print(" Dog Bark")

class Cat:
    def sound(self):
        print("Cat Meow")

d = Dog()
d.sound()

c = Cat()
c.sound()
```

# Abstraction (Show Only What Is Needed)
Hiding internal details and showing only useful features.   
e.g(Car) --> You use steering, brake, accelerator and You don’t know engine working

```python
from abc import ABC, abstractmethod

# Abstract class
class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

# Concrete class 1
class UPI(Payment):
    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI")

# Concrete class 2
class Card(Payment):
    def pay(self, amount):
        print(f"Paid ₹{amount} using Card")

# Using abstraction
p1 = UPI()
p2 = Card()

p1.pay(500)
p2.pay(1000)
```
