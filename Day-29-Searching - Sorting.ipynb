```python
 import numpy as np
```

## Sorting

```python
# Sort Values 1D Array

data = np.array([310, 34, 23, 11])

print('Sorted Values :', np.sort(data))
```

```python
# Sort Values 2D Array

data = np.array([
    
    [310, 34, 23, 11],
    [210, 45, 56, 13]

])


print('Row Wise Sorted Values  :\n', np.sort(data))
print('\n')
print('Column Wise Sorted Values  :\n', np.sort(data, axis=0))


# 0 looks like an egg falling down (Rows).
# 1 looks like a person walking across a bridge (Columns)
```

```python
# Argument Sort (Provide Index No Where Sort Values)
data = np.array([30, 10, 20])

print('Argument Sort :', np.argsort(data))
```

## Searching

```python
# Find Index No In data
a = np.array([10, 20, 30, 40, 50])
print(np.where(a==30))
```

```python
# Find Index No In data Where Satisfy Condition
a = np.array([10, 15, 20, 25, 30])

print(np.where(a % 2 ==0))

# -------------------------------------------------------------
# Find Index No In data Where Satisfy Condition if Not Exist
a = np.array([10, 15, 20, 25, 30])

print(np.where(a % 2 ==60))
```

```python
# Where To Put This No That Array Remain Sorted After Add in All Elements Provide Index Number
print(np.searchsorted(a, 24))
```
