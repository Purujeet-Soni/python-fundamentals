```python
# Creating New File by X (use only files that not exists in system)
new_file = open("New_file.txt", "x")
new_file.write("Hello World")
new_file.close()
```

```python
# Write on New File ( That "w" Delete old Data )  
new_file = open("New_file.txt", "w")
new_file.write("Haa")
new_file.close()
```

```python
# Read New File 
new_file = open("New_file.txt", "r")
print(new_file.read())
new_file.close()
```

```python
# Append New File ( That "a" Add At Last on New File Data )
new_file = open("New_file.txt", "a")
new_file.write(" Hello India")
new_file.close()
```

```python
# Read New File 
new_file = open("New_file.txt", "r")
print(new_file.read())
new_file.close()
```

```python
# Read New File " with " mode no required to close() file 
with open("New_file.txt", "r") as new_file:
    print(new_file.read())
```

```python
# Remove file by OS 
import os 
os.remove("New_file.txt")
```

```python
# r+ mode (Read + Write)
f = open("New_file.txt", "r+")
print(f.read())
f.write("Empty")
f.close()
```

```python
# w+ mode (Write + Read)
f = open("New_file.txt", "w+")
f.write(" Empty New File")
f.seek(0)                        # Move Pointer Back to Start 
print(f.read())
f.close()
```

```python
# a+ mode (Append + Read)
f = open("New_file.txt", "a+")
f.write(" Not Empty Now")
f.seek(0)
print(f.read())
f.close()
```

```python
# rb+ (Read + Write Binary)
f = open(r"E:\Other\DHARA\aastha.jpg", "rb+")
data = f.read(10)
f.write(b"\x00\x01")
f.close()
print(data)
```

```python
# wb+ (Write + Read Binary)
f = open("file.bin", "wb+")
f.write(b"ABC")
f.seek(0)
print(f.read())
f.close()
```

```python
# ab+ (Append + Read Binary)
f = open("log.bin", "ab+")
f.write(b"NEW")
f.seek(0)
print(f.read())
f.close()
```
