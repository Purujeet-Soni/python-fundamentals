# List

```python
# LIST PROPERTIES:
# 1. Ordered (maintains insertion order)
# 2. Mutable (can change values)
# 3. Allows duplicate values
# 4. Indexed
# 5. Can store different data types

my_list = [10, 20, 20, "Python", 3.5]

my_list[1] = 99        # mutable: value can be changed
my_list.append(50)    # adding new element

print(my_list)
```

# Tuple

```python
# TUPLE PROPERTIES:
# 1. Ordered
# 2. Immutable (cannot change values)
# 3. Allows duplicate values
# 4. Indexed
# 5. Faster than list (read-only)

my_tuple = (10, 20, 20, "Python", 3.5)

# my_tuple[1] = 99     # ERROR: tuple is immutable

print(my_tuple)
```

# Sets

```python
# SET PROPERTIES:
# 1. Unordered (no fixed order)
# 2. Mutable (can add/remove elements)
# 3. Does NOT allow duplicate values
# 4. Not indexed
# 5. Used for unique values

my_set = {10, 20, 20, 30, "Python"}

my_set.add(40)        # adding element
my_set.remove(10)     # removing element
print(my_set)
```

# Dictionary

```python
# DICTIONARY PROPERTIES:
# 1. Ordered (Python 3.7+)
# 2. Mutable
# 3. Stores data in key : value pairs
# 4. Keys must be unique
# 5. Values can be duplicated

my_dict = {
    "name": "Rahul",
    "age": 20,
    "course": "Python",
    "list_power": my_list
}

my_dict["age"] = 21        # mutable: value updated
my_dict["city"] = "Delhi" # new key-value added

print(my_dict)
print()
print("Keys:\t", my_dict.keys()) # Print Values Left Side Things
print()
print("Values:\t", my_dict.values()) # Print Values Right Side Things
print()
print("Key : Values Pair :\t", my_dict.items())
```
