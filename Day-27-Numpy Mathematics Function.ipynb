```python
import numpy as np
```

```python
a = np.array([10, 20, 30])
b = np.array([2, 5, 3])
```

```python
# Addition of A and B
print('Addition :', a+b)

# Subtraction of A and B
print('Subtraction :', a-b)

# Multiple of A and B
print('Multiple :', a*b)

# Division of A and B
print('Division :', a/b)
```

```python
x = np.array([1, 4, 9, 16])

# Power 
print('Power by 2 :', np.power(x, 2))

# Square Root
print('Square Root :', np.sqrt(x))
```

```python
# Trignometric Function
Angles = np.array([0, np.pi/2, np.pi]) # pi = 180 Degree

print('Sin :', np.sin(Angles))
print('Cos :', np.cos(Angles))
print('Tan :', np.tan(Angles))
```

```python
# Logarithmic Functions
a = np.array([1, 10, 100])

p = np.log(a)              # Log to Base e = approx 2.718. 
print('Natural Logarithmic: ', p)

q = np.log10(a)            # Log to Base 10
print('\nLog 10: ', q)

r = np.exp(a)              # opposite (inverse) of np.log
print('\nExponential Log: ', r) 

s = np.log1p(a)            # Use Where Negative Or Near Zeror Values Present In Data
print('\nlog1p = Log (x + 1): ', s) 

t = np.expm1(a)            # opposite (inverse) of np.log1p(a)
print('\ne^y - 1: ', t)
```

```python
# Floor - Ceil - Round
x = np.array([1.2, 2.7, 3.5, 4.9])
print('Array     :', x, '\n')

print('Ceil      :', np.ceil(x))      # Next Value
print('Floor     :', np.floor(x))     # Previous Value
print('Round Off :', np.round(x))     # Less then .5 Then Previous Else Next
```

```python
# Basic Statistics
data = np.array([10, 20, 30, 40])
print('Data               :', data)

print('\n')

print('Sum                :', np.sum(data))
print('Sum Mean           :', np.mean(data))
print('Minimum            :', np.min(data))
print('Maximum            :', np.max(data))
print('Standard Deviation :', np.std(data))
print('Varience           :', np.var(data))

```

```python
# Advanced Fucntion (abs, Sign)
array = np.array([-10, -5, 0, 5, 10])
print('Array    : ', array)

# abs (Absolute)
print('Absolute :   ', np.abs(array))  # Remove The Sign Keep The Magnitude

# sign (Sign)
print('Sign     :   ', np.sign(array)) # Provide only Sign
```
