### Library

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
```

### Data

```python
df = [10, 20, 15, 30, 45, 100]
data = pd.DataFrame({
    'grp':['A', 'A', 'B', 'B'], 'val':[10, 20, 15, 30]
})
```

### By Plt

#### Simple Box Plot

```python
plt.boxplot(df)
plt.show()
```

#### Left To Right 

```python
plt.boxplot(df, vert=False)
plt.show()
```

### By Seaborn

#### Simple Box Plot

```python
sns.boxplot(data=data, x='grp', y='val')
plt.show()
```

#### Diffrent Color Chart

```python
sns.boxplot(data=data, x='grp', y='val', hue='grp')
plt.show()
```
