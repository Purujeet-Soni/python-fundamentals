```python
import pandas as pd
```

```python
df = pd.read_csv(r"E:\DATA FOR TEST\Tele Customer churn\WA_Fn-UseC_-Telco-Customer-Churn _1.csv")
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors = 'coerce')
df.info()
```

```python
# Gender Group by Total Charges by Sum 
df.groupby('gender')["TotalCharges"].sum()
```

```python
# Gender Group by Total Charges by mean
df.groupby('gender')["TotalCharges"].mean()
```

```python
# Gender Group by Total Charges by Median
df.groupby('gender')["TotalCharges"].median()
```

```python
# Gender Group by Total Charges by Minimum
df.groupby('gender')["TotalCharges"].min()
```

```python
# Gender Group by Total Charges by Maximum
df.groupby('gender')["TotalCharges"].max()
```

```python
# Gender Group by Count of All Variable
df.groupby('gender').count()
```

```python
# tenure Group by Count of All Variable
df.groupby('tenure').count()
```
