```python
import pandas as pd
```

```python
df = pd.read_csv(r"E:\DATA FOR TEST\Tele Customer churn\WA_Fn-UseC_-Telco-Customer-Churn _1.csv")
print(df.columns)
df.head(1)
```

```python
# find Null Values
# df.isna()
df.isna().sum() # or df.isnull().sum()
```

```python
# Find Not Null
# df.notna()
df.notna().sum()
```

```python
# Find Null Value In Specific Columns
df["gender"].isna().sum()
```

```python
# Find Null Value In Specific Columns
df["gender"].notna().sum()
```

```python
# Fill all Null Value With 0
df.fillna(0)
```

```python
# Fill Column all Null Value With 0
df['Columns Name'].fillna(0)
```

```python
# Fill all Null Value With mean of Total Charges
df.fillna(df["Total Charges"].mean())
```

```python
# Fill Column ALl Null Value With There Mean
df['column'].fillna(df['column'].mean())
```

```python
# Fill Column ALl Null Value With There Median
df['column'].fillna(df['column'].Median())
```

```python
# Fill Column ALl Null Value With There Mode
df['column'].fillna(df['column'].Mode()[0])
```

```python
# Fill Column ALl Null Value With There Minimum Values
df['column'].fillna(df['column'].min())
```

```python
# Fill Column ALl Null Value With There Maximum Values
df['column'].fillna(df['column'].max())
```

```python
# Fill Column All Null Value With There Forward Values Fill
df['column'].fillna(method = 'ffill')
```

```python
# Fill Column All Null Value With There Backward Values Fill
df['column'].fillna(method = 'bfill')
```

```python
# Drop All Null Values
df.dropna()
```

```python
# Column Drop All Null Values
df['column'].dropna()
```
