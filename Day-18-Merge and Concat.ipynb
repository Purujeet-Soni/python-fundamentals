```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "EmployeeID": [101.0, 102.0, 103.0, 104.0, 105.0, 106.0],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Age": [25, 30, 40, 40, 29, 40],
    "Salary": [50000.0, 45000.0, 60000.0, np.nan, 48000.0, 70000.0],
    "Experience": [2.0, 5.0, 4.0, 10.0, np.nan, 7.0]
})
df
```

```python
Manager = pd.DataFrame({
    "Department": ["IT", "HR", "Finance"],
    "Manager": ["A", "B", "C"]
})
Manager
```

```python
# Merge Both Data on Department (Add Columns)
Merge_data = pd.merge(df, Manager, on='Department')
Merge_data
```

```python
# Merge Both Data on Same (Add Columns and Row Cross)
Merge_data_same = pd.merge(df, df, on='Department')
Merge_data_same
```

```python
# Concate Data (Add Rows)
Concat_data = pd.concat([df, df])
Concat_data
```

```python
# Concate Data on Diffrent Data shape(Add Rows)
Concat_data = pd.concat([df, Manager])
Concat_data
```
