## Library

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Data
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

df = pd.DataFrame({'Days':[1,2,3,4], 'Sales':[1000, 2000, 1500, 2500]})
```

## Line Plot By plt

#### Simple Plot the Line Plot

```python
plt.plot(x, y)
plt.show()
```

#### With Red Color and -- Pattern

```python
plt.plot(x, y, 'r--')
plt.show()
```

#### With Marker

```python
plt.plot(x, y, marker='o')
plt.show()
```

#### Customize

```python
# Same
plt.figure(figsize=(8, 3))
plt.plot(x, y, marker='*', linestyle='-.', color ='purple', alpha=0.99) 
plt.grid(alpha=0.14)
plt.xticks([x for x in np.linspace(1, 4.5, 15)])
plt.yticks(range(10, 29, 2))
plt.show()
```

## Line Plot By Seaborn

#### Simple Line Plot By Seaborn

```python
sns.lineplot(data=df, x='Days', y='Sales')
plt.show()
```

#### Red Color and -- Pattern

```python
sns.lineplot(data=df, x='Days', y='Sales', linestyle='--', color='red')
plt.show()
```

#### With Marker

```python
sns.lineplot(data=df, x='Days', y='Sales', marker='o')
plt.show()
```

#### Customize

```python
# Same
plt.figure(figsize=(8, 3))
sns.lineplot(x=df['Days'], y=df['Sales'], linestyle='-.', color ='purple', alpha=0.78, marker='d', markerfacecolor="green", markeredgecolor="black", markersize=8)
plt.grid(alpha=0.14)
plt.xticks([x for x in np.linspace(1, 4.5, 15)])
plt.yticks([x for x in np.linspace(1000, 2600, 9)])
plt.show()
```
