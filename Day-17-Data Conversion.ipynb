```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv(r"E:\DATA FOR TEST\50 Startups\50_Startups.csv")
data.head(2)
```

```python
data.info()
```

```python
# Convert Float Value to Interger 
data['Profit'] = data['Profit'].astype('int')
```

```python
#  Canont Convert Null Values Columns Data Type
data['Profit'] = data['Profit'].fillna(data['Profit'].mean())


# Same Code Above

# Convert Float Value to Interger 
data['Profit'] = data['Profit'].astype('int')
```

```python
data.info()
```

```python
# Again Convert integer to float
data['Profit'] = data['Profit'].astype('float')
print(data.info())
data.head(2)
```

```python
# Due to Changes After Point Values Are Disabel so we use Round
data['Profit'] = data['Profit'].round(2)
data.head(2)
```

```python
# Convert State to Category
data['State'] = data['State'].astype('category')
data.info()
```

```python
# Convert State to Object
data['State'] = data['State'].astype('object')
data.info()
```

```python
# Convert Profit to Object
data['Profit'] = data['Profit'].astype('object')
data.info()
```

```python
# Convert Profit to Object
data['Profit'] = data['Profit'].astype('float64')
data.info()
```
