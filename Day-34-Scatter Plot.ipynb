```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

## Data

```python
x = [1, 2, 3, 4]
y = [10, 15, 13, 20]

data = pd.DataFrame({
    'x':[1, 2, 3, 4],
    'y':[10, 15, 13, 20]
})

df1 = pd.DataFrame({
    'x':[10, 20, 30, 40],
    'y':[11, 16, 14, 21]
})

df2 = pd.DataFrame({
    'x':[50, 80, 90, 40],
    'y':[12, 17, 16, 23]
})
```

## By plt

#### Simple

```python
plt.scatter(x, y)
plt.show()
```

#### Diffrent Color

```python
plt.scatter(x, y, color = 'red')
plt.show()
```

#### Use Diffrent Markers

```python
plt.scatter(x, y, marker = '*')
plt.show()
```

## By Seaborn

#### Simple

```python
sns.scatterplot(data=data, x='x', y='y')
plt.show()
```

#### Diffrent Color

```python
sns.scatterplot(data=data, x='x', y='y', color='green')
plt.show()
```

#### Plot Diffrent Data

```python
sns.scatterplot(data=df1, x='x', y='y', color='gold', edgecolor='orange')
sns.scatterplot(data=df2, x='x', y='y', color='silver')
plt.show()
```

#### Diffrent Size

```python
sns.scatterplot(data=data, x='x', y='y', size='y')
plt.show()
```

#### Customize

```python
sns.scatterplot(data=data, x='x', y='y', size='y', color='red', marker='*', edgecolor='black')
plt.show()
```
