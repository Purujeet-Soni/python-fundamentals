# Operators

```python
a = 10
b = 20

print(f"Addition : {a} , {b} -->", a + b)
print()

print(f"Subtraction : {a} , {b} -->", a - b)
print()

print(f"Multiplication : {a} , {b} -->", a * b)
print()

print(f"Division : {b} , {a} -->", a / b)
print()

print(f"Modulus : {b} , {a} -->", b % a)
print()

print(f"Floor Division : {b} , {a} -->", b // a)
print()
print(f"Power : {a} , {b} -->", a ** b)
```

```python
# Assignment Operators
a = 10

a += 5    # a = a + 5
print("a += 5 -->", a)

a -= 3    # a = a - 3
print("a -= 3 -->", a)

a *= 2    # a = a * 2
print("a *= 2 -->", a)

a /= 4    # a = a / 4
print("a /= 4 -->", a)

a %= 3    # a = a % 3
print("a %= 3 -->", a)
```

```python
# Comparison Operators Example
x = 10
y = 20

print("x == y -->", x == y)   # equal to
print("x != y -->", x != y)   # not equal to
print("x > y -->", x > y)     # greater than
print("x < y -->", x < y)     # less than
print("x >= y -->", x >= y)   # greater than or equal to
print("x <= y -->", x <= y)   # less than or equal to
```

```python
# Comparision Operator
a = 15
b = 10

print("a == b -->", a == b)
print("a != b -->", a != b)
print("a > b -->", a > b)
print("a < b -->", a < b)
print("a >= b -->", a >= b)
print("a <= b -->", a <= b)
```

```python
# Logical Operator
a = 10
b = 20

print("a > 5 and b > 15 -->", a > 5 and b > 15)
print("a > 5 or b < 15 -->", a > 5 or b < 15)
print("not(a > 5) -->", not(a > 5))
```

# Conditional Statements

```python
a, b = 10, 20

# if Statement

if a < b:
    print(f"{a} is Smaller Then {b}")
```

```python
a, b = 20, 10

# if - else Statement

if a < b:
    print(f"{a} is Smaller Then {b}")
else:
    print(f"{a} is Greater Then {b}")
```

```python
a, b = 15, 15

# if - elif - else Statement

if a < b:
    print(f"{a} is Smaller Then {b}")
elif a == b:
    print(f"{a} is Equal to {b}")
else:
    print(f"{a} is Greater Then {b}")
```

```python
# Odd Even Find

num = 13
if num % 2 == 0:
    print(f" Number {num} : Even")
else:
    print(f" Number {num} : Odd")
```
