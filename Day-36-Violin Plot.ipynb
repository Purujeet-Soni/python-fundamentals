### Library

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

### Data

```python
data = {
    'Category' : ['A', 'A', 'A', 'B', 'B', 'B', 'C', 'C', 'C'],
    'Score' : [10, 12, 14, 20, 22, 19, 30, 28, 35]
}

df = pd.DataFrame(data)
```

### By Seaborn

#### Simple Plot

```python
sns.violinplot(data = df, x='Category', y='Score')
plt.show()
```

#### Hue by Category for Diffrenct Color

```python
sns.violinplot(data = df, x='Category', y='Score', hue='Category')
plt.show()
```

#### Hue by Category for Diffrenct Color Customize 

```python
sns.violinplot(data = df, x='Category', y='Score', color='skyblue')
plt.show()
```

#### Horizontal Violin Just Change X and Y

```python
sns.violinplot(data=df, x='Score', y='Category')
plt.show()
```

#### Quartile Changed From Box Plot

```python
sns.violinplot(data=df, x='Category', y='Score', inner='quartile')
plt.show()
```

# Customize

```python
plt.figure(figsize=(13, 4))
sns.violinplot(data=df, x='Score', y='Category', color='gold', inner='quartile')
plt.grid()
plt.xticks(range(1, 41))
plt.show()
```

```python
plt.figure(figsize=(12, 13))
sns.violinplot(data=df, y='Score', x='Category', color='gold', inner='quartile')
plt.grid()
plt.yticks(range(1, 41))
plt.show()
```
