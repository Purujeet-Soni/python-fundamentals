```python
import numpy as np
```

```python
arr = np.array([10, 20, 30, 40])

mask = arr > 25

masked_arr = arr[mask]
masked_arr
```

```python
# Direct

arr = np.array([10, 20, 30, 40, 50, 60, 30, 20, 20, 20, 20, 30, 40, 50, 5])

res = arr[arr > 29]
res
```

```python
# And OR Condition also Apply

# And
arr = np.array([10, 20, 300, 400, 500, 60, 30, 20, 20, 20, 20, 30, 40, 50, 5])

res = arr[(arr > 29) & (arr < 300)]
print(f'Result by &(AND) Operation :\n', res)

# Or
arr = np.array([10, 20, 300, 400, 500, 60, 30, 20, 20, 20, 20, 30, 40, 50, 5])

res = arr[(arr > 29) | (arr < 300)]
print(f'\nResult by |(OR) Operation :\n', res)
```

```python
# Experiment on Same 2D Array it Return 1D Array
arr = np.array([[1, 2, 3, 4], [11, 22, 33, 44]])

res = arr[(arr > 3) & (arr < 44)]
res
```

```python
# Values Assign with Condition
age = np.array([12, 11, 56, 23, 75, 23, 70, 26, 92, 82, 88])
age[(age < 20) & (age > 10)] += 1
```

```python
# Advance Values Assign with Condition
label = np.where(
    (
        (age > 20)
        &
        (age < 75)
    ),
    
    'Young',
    'other'
)

label
```

```python
# Advance Values Assign with Condition
Equality = np.where(
    age < 20, 'Child',
    np.where(age > 75, 'Old', 'Young')
                   )
Equality
```
