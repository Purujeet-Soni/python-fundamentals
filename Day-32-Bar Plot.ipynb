## Library

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

## Data

```python
# For Matplotlib.pyplot
x = ['A', 'B', 'C']
y = [10, 25, 15]

# For Seaborn
df = pd.DataFrame(
    {
        'Cat':['A', 'B', 'C'],
        'Val':[10, 25, 15]
    }
)
```

## Bar Plot by plt

#### Simple Bar Plot

```python
plt.bar(x, y)
plt.show()
```

#### Horizontal Bar Chart

```python
plt.barh(x, y)
plt.show()
```

#### Bar Green Color

```python
plt.bar(x, y, color='green')
plt.plot()
```

#### Customize Bar Chart 

```python
plt.figure(figsize=(6, 4))
plt.bar(x, y, color='gold', edgecolor='orange')
plt.grid(axis='y', alpha=0.23, color='silver')
plt.yticks([x for x in np.linspace(0, 26, 14)])
plt.show()
```

#### Customize Horizontal Bar Chart

```python
plt.figure(figsize=(6, 4))
plt.barh(x, y, color=['Orange', 'gold', 'Silver'], edgecolor='orange', alpha=0.99)
plt.grid(axis='x', alpha=0.13, color='black')
plt.xticks([x for x in np.linspace(0, 26, 14)])
plt.show()
```

## Bar Plot By Seaborn

#### Simple Bar Plot

```python
sns.barplot(x='Cat', y='Val', data=df)
plt.show()
```

#### Horizontal Bar Plot 

```python
sns.barplot(x='Val', y='Cat', data=df, orient='h')
plt.show()
```

#### Diffrent Color

```python
sns.barplot(x='Cat', y='Val', data=df, hue='Cat')
plt.show()
```

#### Customized Bar Chart

```python
plt.figure(figsize=(6, 4))
sns.barplot(data=df, x='Cat', y='Val', edgecolor='Black', hue='Cat', legend=True)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='y', alpha=0.23, color='silver')
plt.yticks([x for x in np.linspace(0, 26, 14)])
plt.show()
```
