```python
import pandas as pd
import numpy as np

data = {
    "EmployeeID": [101, 102, 103, 104, 105, 106],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Age": [25.0, 30.0, np.nan, 40.0, 29.0, np.nan],
    "Salary": [50000.0, 45000.0, 60000.0, np.nan, 48000.0, 70000.0],
    "Experience": [2.0, 5.0, 4.0, 10.0, np.nan, 7.0]
}

df = pd.DataFrame(data)
print(df)
```

```python
df["Join Date"] = pd.to_datetime(["2021-01-01", "2020-05-03", "2019-08-10", "2022-02-01", "2021-07-15", "2018-06-20"])
df
```

```python
a = df.set_index(df["Join Date"])
a.sort_index(inplace = True)
a
```

```python
# Date Time Year 
df["Year"] = df["Join Date"].dt.year
df
```

```python
# Date Time Month
df["Month"] = df["Join Date"].dt.month
df
```

```python
# Date Time Day
df["Day"] = df["Join Date"].dt.day
df
```

```python
# Date Time Current Date Time
df["Current Date Time"] = pd.to_datetime("now")
df
```

```python
# Current Date Time Day
df["Hour"] = df["Current Date Time"].dt.hour
df

# Current Date Time Day
df["Minites"] = df["Current Date Time"].dt.minute
df

# Current Date Time Day
df["Seconds"] = df["Current Date Time"].dt.second
df
```

```python
# Date Time Current Date Time 
df["Current Date Time 1"] = pd.Timestamp.now()
df
```

```python
# Set Current Date Time As Index 
df.set_index(df["Current Date Time 1"])
# df.sort_index()
```

```python
df.to_csv()
```
