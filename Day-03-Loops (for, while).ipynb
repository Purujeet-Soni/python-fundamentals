# For

```python
# For Print 0 to 4 Number
for i in range(5):
    print(i)

```

```python
# For Print 1 to 5 Number
for i in range(1, 6):
    print(i)
```

```python
# Inverse Print 1 to 5 Number by -1
for i in range(5, 0, -1):
    print(i) 

print()

# Inverse Print 1 to 5 Number by Reversed
for i in reversed(range(1, 6)):
    print(i) 
```

```python
# For Print 1 to 5 Number Horizontally
for i in range(1, 6):
    print(i, end="")
```

```python
# Countinue(Skip) on 5
for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

```python
# Break(Stop) Statement on 5
for i in range(1, 11):
    if i == 5:
        break
    print(i)
```

# While

```python
# While Print N Not Equal to 10
n = 1
while n <= 10:
    print(n)
    n += 1
```

```python
# Inverse While Print N Not Equal to 10
n = 10
while n >= 1:
    print(n)
    n -= 1
```
