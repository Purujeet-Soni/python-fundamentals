# Before

```python
# string Data type
x = "100"
y = "3.5"
z = "True"

print(x, type(x))
print(y, type(y))
print(z, type(z))
```

# After

```python
# integer (after type conversion)
a = int(x)

# float (after type conversion)
b = float(y)

# boolean (after type conversion)
c = bool(z)

print(a, type(a))
print(b, type(b))
print(c, type(c))
```

```python
# integer
p = 25

# float (type conversion from int)
q = float(p)

# string (type conversion from int)
r = str(p)

print(p, type(p))
print(q, type(q))
print(r, type(r))
```

# Input Function

```python
# For String
name = input("Please Enter Your Name :\t")
print(name.title(), type(name))

print()

# For Float
Float = float(input("Please Enter Any Float Value:\t"))
print(Float, type(Float))
```
