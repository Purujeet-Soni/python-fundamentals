```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Define the categories (The companies from your image)
categories = ['Newgen Software', 'Data Patterns', 'Apar Industries', 'Praj Industries', 'Safari Ind.']
N = len(categories)

# 2. Define the data for the two polygons
# (Using placeholder quantitative data scaled from 0 to 100)
blue_series = [90, 85, 75, 95, 80]
orange_series = [30, 25, 35, 20, 25]

# 3. Close the loop! Append the first value to the end of the arrays
blue_series += blue_series[:1]
orange_series += orange_series[:1]

# 4. Calculate the angle for each category on the polar axis
angles = [n / float(N) * 2 * np.pi for n in range(N)]
angles += angles[:1]

# 5. Initialize the plot with a polar projection
fig, ax = plt.subplots(figsize=(7, 7), subplot_kw=dict(polar=True))

# 6. Plot the Blue outer series
ax.plot(angles, blue_series, color='#1f3b8c', linewidth=2, linestyle='solid', marker='o')
ax.fill(angles, blue_series, color='#8ba0cd', alpha=0.4) # Alpha controls transparency

# 7. Plot the Orange inner series
ax.plot(angles, orange_series, color='#f29c11', linewidth=2, linestyle='solid', marker='o')
ax.fill(angles, orange_series, color='#f5c77e', alpha=0.6)

# 8. Add the labels for each axis
plt.xticks(angles[:-1], categories, size=10, color='dimgrey')

# 9. Clean up the grid (Optional: hiding the radial y-axis labels for a cleaner look)
ax.set_yticklabels([])
ax.grid(color='lightgrey', linestyle='-', linewidth=1)

# Display the chart
plt.show()
```
